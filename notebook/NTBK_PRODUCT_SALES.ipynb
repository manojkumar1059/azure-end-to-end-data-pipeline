```python
df = spark.read.format("csv").option("header", "true").option("inferSchema", "true").load("/mnt/raw_data/product_sales.csv")
display(df)
```

```python
df_no_nulls = df.dropna()
display(df_no_nulls)
```

```python
from pyspark.sql import functions as F

duplicates_df = df.groupBy(df.columns).count().filter(F.col("count") > 1)
display(duplicates_df)
```

```python
df_clean = df.dropDuplicates()
display(df_clean)
```

```python
from pyspark.sql.functions import col, round, month, year, quarter

df = (
    df.withColumn("Profit_Margin", round((col(" Profit ") / col(" Revenue ")) * 100, 2))
      .withColumn("Month", month(col("Order_Date")))
      .withColumn("Year", year(col("Order_Date")))
      .withColumn("Quarter", quarter(col("Order_Date")))
)

display(df.limit(10))
```

```python
# Remove leading/trailing spaces from all column names
df = df.toDF(*[c.strip() for c in df.columns])

from pyspark.sql.functions import sum, avg, round

sales_region = (
    df.groupBy("Region")
      .agg(
          round(sum("Revenue"), 2).alias("Total_Revenue"),
          round(sum("Profit"), 2).alias("Total_Profit"),
          round(avg("Profit_Margin"), 2).alias("Avg_Profit_Margin")
      )
)

display(sales_region)
```

```python
sales_category = df.groupBy("Category", "Sub_Category") \
    .agg(
        round(sum("Revenue"), 2).alias("Total_Revenue"),
        round(sum("Profit"), 2).alias("Total_Profit"),
        round(avg("Unit_Price"), 2).alias("Avg_Unit_Price")
    )
display(sales_category)

```

```python
top_products = df.groupBy("Product_Name") \
    .agg(
        round(sum("Revenue"), 2).alias("Total_Revenue"),
        round(sum("Profit"), 2).alias("Total_Profit")
    ).orderBy(col("Total_Revenue").desc())
display(top_products)
```

```python
monthly_trend = df.groupBy("Year", "Month") \
    .agg(
        round(sum("Revenue"), 2).alias("Total_Revenue"),
        round(sum("Profit"), 2).alias("Total_Profit")
    ).orderBy("Year", "Month")
display(monthly_trend)
```

```python
state_perf = df.groupBy("State") \
    .agg(
        round(sum("Revenue"), 2).alias("Total_Revenue"),
        round(sum("Profit"), 2).alias("Total_Profit"),
        round(avg("Profit_Margin"), 2).alias("Avg_Profit_Margin")
    )
display(state_perf)
```
